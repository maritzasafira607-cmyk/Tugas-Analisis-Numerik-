# Volume Kedua Lambung Katamaran (Integrasi Numerik)

**Asumsi**
- 1 kotak grid = 5 + 5 = 10 cm.
- Lantai datar (tanpa keel), tampak depan persegi panjang, jadi luas penampang $A(x) = lebar(x) \times tinggi(x)$.
- Reserve buoyancy tidak dihitung.
- Volume satu lambung: $V = \int_{x_0}^{x_1} A(x)\,dx$, lalu dikali 2 untuk kedua lambung.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import trapezoid, simpson

# Titik data hasil pembacaan grid (cm). x diukur dari ujung kiri gambar.
x_data      = np.array([20, 44, 130, 181])
lebar_data  = np.array([17, 20, 20, 10])
tinggi_data = np.array([16.6, 26, 26, 26])

def lebar(x):  return np.interp(x, x_data, lebar_data)
def tinggi(x): return np.interp(x, x_data, tinggi_data)
def luas(x):   return lebar(x) * tinggi(x)
```

## Metode 1: Trapesium pada titik data (3 segmen)

```python
V_trap_kasar = trapezoid(luas(x_data), x_data)
print(f"Trapesium 3 segmen: {V_trap_kasar:,.0f} cm3 = {V_trap_kasar/1000:.1f} L")
```

## Metode 2: Simpson 1/3 per segmen
$A(x)$ adalah hasil kali dua fungsi linear (kuadratik), sehingga Simpson menghasilkan nilai eksak untuk tiap segmen.

```python
V_simp = 0
for a, b in zip(x_data[:-1], x_data[1:]):
    xs = np.linspace(a, b, 3)          # 2 subinterval (jumlah genap)
    V_simp += simpson(luas(xs), x=xs)
print(f"Simpson per segmen: {V_simp:,.0f} cm3 = {V_simp/1000:.1f} L")
```

## Metode 3: Trapesium dengan banyak titik (uji konvergensi)

```python
for n in [4, 10, 50, 200, 1000]:
    xs = np.linspace(x_data[0], x_data[-1], n)
    V = trapezoid(luas(xs), xs)
    print(f"n = {n:5d}  ->  V = {V:,.1f} cm3   (galat vs Simpson: {abs(V - V_simp):,.1f} cm3)")
```

## Hasil akhir (kedua lambung)

```python
V_total = 2 * V_simp
print(f"Satu lambung  : {V_simp/1000:.1f} L")
print(f"Kedua lambung : {V_total:,.0f} cm3 = {V_total/1000:.1f} L = {V_total/1e6:.3f} m3")
```

## Grafik

```python
xs = np.linspace(x_data[0], x_data[-1], 400)
fig, ax = plt.subplots(1, 2, figsize=(11, 3.8))
ax[0].plot(xs, lebar(xs), label="lebar")
ax[0].plot(xs, tinggi(xs), label="tinggi")
ax[0].set_xlabel("x (cm)"); ax[0].set_ylabel("cm"); ax[0].legend(); ax[0].set_title("Profil lambung")
ax[1].fill_between(xs, luas(xs), alpha=0.4)
ax[1].set_xlabel("x (cm)"); ax[1].set_ylabel("A(x) (cm2)"); ax[1].set_title("Luas penampang (area = volume)")
plt.tight_layout(); plt.show()
```
